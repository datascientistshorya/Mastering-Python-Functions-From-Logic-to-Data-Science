#### Q1. Data cleaning

Write a function clean_values(data) that removes None, empty strings (""), and strings containing only whitespace from a list.

Input:

data = [10, None, "hello", "", "  ", 25, "Python", None]

```python
data = [10, None, "hello", "", "  ", 25, "Python", None]

def clean_values(data):
    clean=[]
    for i in data:
        if i is None:
            continue
        elif isinstance(i, str) and i.strip() == "":
            continue
        else:
            clean.append(i)
    return clean

print(clean_values(data))
```

#### Q2. Data validation

Write a function validate_scores(scores) that returns only valid numeric scores between 0 and 100, inclusive.

Input:

scores = [45, 105, -5, 80, None, 67, 100]

Expected output:

[45, 80, 67, 100]

Assume the input contains numbers or None

```python
scores = [45, 105, -5, 80, None, 67, 100]

def  validate_scores(scores):
    valid=[]
    for i in scores:
        if i is None:
            continue
        if i>=0 and i<=100:
            valid.append(i)
    return valid

print(validate_scores(scores))


```

#### Q3. Feature transformation

Write a function normalize_min_max(values) that scales numeric values between 0 and 1 using min-max normalization.

Formula:

x scaled = x=min(x)/max(x)−min(x)

Requirements:

Return an empty list for empty input.

Return a list of zeros when all values are identical.

```python
values=[1,34,56,34,89,56]

def normalize_min_max(values):
    if values==[]:
        return values
    if len(set(values)) == 1:
        return [0] * len(values)
    return values
    min_i=min(values)
    max_i=max(values)
    normal=[]
    for i in values:
        normal.append((i-min_i)/(max_i-min_i))
    return(normal)

print(normalize_min_max(values))


```

#### Q4. Reusable filtering

Write a function filter_adults(people) that accepts a list of dictionaries and returns only people aged 18 or above.

Input:

people = [
    {"name": "Aman", "age": 17},
    {"name": "Riya", "age": 22},
    {"name": "Kabir", "age": 16},
    {"name": "Neha", "age": 30}
]

```python
people = [
    {"name": "Aman", "age": 17},
    {"name": "Riya", "age": 22},
    {"name": "Kabir", "age": 16},
    {"name": "Neha", "age": 30}
]

def filter_adults(people):
    adults=[]
    for i in people:
        if i['age']>=18:
            adults.append(i)
    return adults

print(filter_adults(people))

```

#### Q5. Missing-value replacement

Write a function replace_missing(data, replacement) that replaces every None value with the supplied replacement.

Input:

data = [10, None, 30, None, 50]
replacement = 0

```python
data = [10, None, 30, None, 50]
replacement = 0

def  replace_missing(data, replacement):
    clean=[]
    for i in data:
        if i is not None:
            clean.append(i)
        else:
            clean.append(replacement)
    return clean

print(replace_missing(data, replacement))
        
        

```

#### Q6. Outlier filtering

Write a function remove_outliers(data, lower, upper) that returns values within the inclusive range from lower to upper.

Input:

data = [12, 15, 18, 120, -10, 22, 25]
lower = 10
upper = 30

```python
data = [12, 15, 18, 120, -10, 22, 25]
lower = 10
upper = 30

def remove_outliers(data, lower, upper):
    clean=[]
    for i in data:
        if i< lower or i>upper:
            continue
        else:
            clean.append(i)
    return clean

print(remove_outliers(data, lower, upper))
```

#### Q7. Categorical encoding

Write a function encode_categories(categories) that assigns a unique integer to each distinct category, starting from zero, in the order each category first appears.

Input:

categories = ["red", "blue", "red", "green", "blue", "yellow"]

Expected output:

[0, 1, 0, 2, 1, 3]

```python
categories = ["red", "blue", "red", "green", "blue", "yellow"]

def encode_categories(categories):
    mapping = {}
    encoded = []

    for i in categories:
        if i not in mapping:
            mapping[i] = len(mapping)

        encoded.append(mapping[i])

    return encoded

print(encode_categories(categories))
```

#### Q8. Dataset summary

Write a function summarize_data(data) that returns a dictionary containing:

count: number of values

minimum: smallest value

maximum: largest value

average: arithmetic mean

Requirements:

Do not use the built-in sum().

Return zeros for all four fields when the list is empty.

```python
data=[34,78,45,90,34,-12,7,78,89]

def summarize_data(data):
    l=len(data)
    total=0
    for i in data:
        total+=i
    summary={}
    summary['total_item']=l
    summary['max_item']=max(data)
    summary['min_item']=min(data)
    summary['Average']=total/l
    return summary

print(summarize_data(data))

```

#### Q9. Feature engineering

Write a function create_features(records) that takes a list of dictionaries containing price and quantity and adds a total_value field equal to price * quantity.

Input:

records = [
    {"price": 100, "quantity": 2},
    {"price": 50, "quantity": 5},
    {"price": 200, "quantity": 1}
]

```python
records = [
    {"price": 100, "quantity": 2},
    {"price": 50, "quantity": 5},
    {"price": 200, "quantity": 1}
]

def create_features(records):
    for i in records:
        i['total_value']=i['price']*i['quantity']
    return records

print(create_features(records))
```

#### Q10. End-to-end data preprocessing

Write a function preprocess_dataset(data) that performs these steps:

Removes None values.

Keeps only numeric values greater than or equal to zero.

Calculates the average of the remaining values without using sum().

Returns a dictionary containing the cleaned list, its count, and its average.

If no valid values remain, returns an empty list, count 0, and average 0.

Input:

data = [10, None, -5, 20, 30, None, 40, -2]

```python
data = [10, None, -5, 20, 30, None, 40, -2]

def preprocess_dataset(data):
    cleaned=[]
    processed={}
    for i in data:
        if i is None or i<0:
            continue
        else:
            cleaned.append(i)
    if len(cleaned)==0:
        processed['cleaned']=cleaned
        processed['total_ items']=0
        processed['Average']=0
        return processed
        
    total=0
    for i in cleaned:
        total+=i
    processed['cleaned']=cleaned
    processed['total_ items']=len(cleaned)
    processed['Average']=total/len(cleaned)
    return processed

print(preprocess_dataset(data))
        
```
