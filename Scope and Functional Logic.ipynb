#### Q6. Function composition

Write a function calculate_average(numbers) that calls a second function, calculate_total(numbers), to calculate the sum before computing the average.

Requirements:

Return 0 for an empty list.

Do not use the built-in sum() function.

```python
def calculate_average(numbers):
    l=len(numbers)
    total=calculate_total(numbers)
    print('Avg is ', total/l)

def calculate_total(numbers):
    total=0
    for i in numbers:
        total+=i
    return total

numbers=[1,2,3,4,5]
calculate_average(numbers)
```

#### Q10. Mini data-analysis challenge

Write a function named analyze_scores(scores) that:

Accepts a list of numeric scores.

Calculates the average without using sum().

Counts how many scores are at least 70.

Returns a dictionary containing average and high_score_count.

Returns {"average": 0, "high_score_count": 0} if the list is empty.

Test it with:

scores = [45, 72, 88, 65, 91, 70]

```python
def analyze_scores(scores):
    high=0
    total=0
    l=len(scores)
    for i in scores:
        total+=i
        if i>=70:
            high+=1

    cal={}
    cal['Avg']=total/l
    cal['high_score_count']=high
    return cal



scores = [45, 72, 88, 65, 91, 70]
print(analyze_scores(scores))
```
