#### Q6. Lists as arguments

Write a function that accepts a list of numbers and returns their average. If the list is empty, return 0.

```python
def avg(n):
    if len(n)==0:
        return 'empty list'
    total=0
    for i in n:
        total+=i
    avg_=total/len(n)
    return avg_

sales=[12,23,34,45]
result=avg(sales)
print(result)
```

#### Q8. Function composition

Write two functions:

clean_numbers(values) returns a new list containing only positive numbers.

calculate_average(values) returns the average of the supplied numbers, or 0 if the list is empty.

Use the first function's output as the input to the second.

```python
def clean_numbers(n):
    l=len(n)
    if l==0:
        return 'Empty list'
    positive=[]
    for i in n:
        if i>-1:
            positive.append(i)
    return positive

def avg(n):
    l=len(n)
    if l==0:
        return 'Empty list'
    total=0
    for i in n:
        total+=i
    return total/l

num=[-9,1,2,3,0,-8]
cleaned=clean_numbers(num)
avg_=avg(num)
print(cleaned,avg_)
```

#### Q10. Machine Learning application

Write a function calculate_metrics(actual, predicted) that accepts two equally sized, non-empty lists of numeric values and returns a dictionary containing:

MAE: Mean Absolute Error

MSE: Mean Squared Error

If the lists have different lengths or either list is empty, return an empty dictionary.

```python
def metrics(act,pre):
    result={}
    if len(act)!= len(pre):
            return result
    a_err=[]
    s_err=[]
   
    for i,j in zip(act,pre):
        a_err.append(i-j)
        s_err.append((i-j)**2)
        
    result['MAE']=sum(a_err)/5
    result['MSE']=sum(s_err)/5
    return result
        
        
        
    
predicted=[45,56,34,78,89]
actual=[50,45,40,80,93]
res=metrics(actual,predicted)
print(res)
```

#### Q10. Data Science challenge

Write a function calculate_summary(data, *metrics, **options) that:

Accepts a list of numeric values.

Accepts requested metrics such as "sum", "mean", "min", and "max" through *metrics.

Accepts an optional keyword argument round_digits through **options, defaulting to 2.

Returns a dictionary containing only the requested metrics.

Returns an empty dictionary if the input list is empty.

```python
def calculate_summary(data, *metrics, **options):
    if not data:
        return {}

    round_digits = options.get("round_digits", 2)
    result = {}

    for metric in metrics:
        if metric == "sum":
            result["sum"] = round(sum(data), round_digits)

        elif metric == "mean":
            result["mean"] = round(sum(data) / len(data), round_digits)

        elif metric == "min":
            result["min"] = round(min(data), round_digits)

        elif metric == "max":
            result["max"] = round(max(data), round_digits)

    return result


data = [3.333,6.6666,1.2543]

print(calculate_summary(data, "sum", "mean", "max"))
print(calculate_summary(data, "min", "mean", round_digits=3))
print(calculate_summary([] , "sum", "max"))
```
