#### Q1. Revenue after tax
prices = [100, 250, 500, 1000, 2000]



Use map() and a lambda to calculate prices after adding 18% tax.

```python
prices = [100, 250, 500, 1000, 2000]
tax_on_prices=lambda x: x*0.18
tax=(list(map(tax_on_prices,prices)))
revenue=[]
for i,j in zip(prices,tax):
    revenue.append(i-j)
print(revenue)
```

#### Q6. Sort products by conversion rate
products = [    {"name": "Laptop", "visitors": 1000, "purchases": 80},    {"name": "Mouse", "visitors": 2000, "purchases": 300},    {"name": "Monitor", "visitors": 800, "purchases": 96},    {"name": "Keyboard", "visitors": 1500, "purchases": 90}]



Sort products by conversion rate from highest to lowest, using sorted() and a lambda.

```python
products = [{"name": "Laptop", "visitors": 1000, "purchases": 80},
            {"name": "Mouse", "visitors": 2000, "purchases": 300},
            {"name": "Monitor", "visitors": 800, "purchases": 96},
            {"name": "Keyboard", "visitors": 1500, "purchases": 90}]
for i in products:
    i['cvr']=(i['purchases']/i['visitors'])*100
print(list(sorted(products, key= lambda products:products['cvr'],reverse=True)))
```

#### Q8. Mini web analytics challenge
campaigns = [    {"channel": "Paid Search", "sessions": 5000, "purchases": 250},    {"channel": "Organic", "sessions": 8000, "purchases": 560},    {"channel": "Social", "sessions": 6000, "purchases": 180},    {"channel": "Email", "sessions": 2000, "purchases": 220},    {"channel": "Referral", "sessions": 3000, "purchases": 150}]



Perform these three operations:
1. Use filter() to keep campaigns with at least 200 purchases.
2. Use map() to extract their conversion rates as percentages.
3. Use sorted() to rank the qualifying campaigns by conversion rate, highest first.
Your final output should contain the channel name and conversion rate for each qualifying campaign.

```python
campaigns = [{"channel": "Paid Search", "sessions": 5000, "purchases": 250},
             {"channel": "Organic", "sessions": 8000, "purchases": 560},
             {"channel": "Social", "sessions": 6000, "purchases": 180},
             {"channel": "Email", "sessions": 2000, "purchases": 220},
             {"channel": "Referral", "sessions": 3000, "purchases": 150}]
data=list(filter(lambda campaigns:campaigns['purchases']>=200,campaigns))
list(map(
    lambda c: c.update({
        "cvr": (c["purchases"] / c["sessions"]) * 100
    }),
    data
))
final=sorted(data, key= lambda data:data['cvr'],reverse=True)
for i in final:
    print(i['channel'],i['cvr'])
    
```
